```python
#!/usr/bin/env python
# coding: utf-8

# =============================================================================
# EXPLAINABLE AI IN CREDIT RISK MANAGEMENT
# Extended & Improved Version — Based on arXiv:2103.00949
# (Misheva, Hirsa, Osterrieder et al., 2021)
#
# DATASET  : German Credit (GermanCredit.csv — upload this one file)
# TARGET   : credit_risk  (1 = Good, 0 = Bad)
#
# HOW TO USE IN GOOGLE COLAB
# ─────────────────────────────────────────────────────────────────────────────
#   Step 1: Upload GermanCredit.csv using the Files panel (left sidebar)
#   Step 2: Run Cell 1 (installs libraries, ~3 min)
#   Step 3: Run all remaining cells top-to-bottom
#
# WHAT THE ORIGINAL PAPER DID (arXiv:2103.00949)
# ─────────────────────────────────────────────────────────────────────────────
#   • Dataset       : Lending Club (2.2M rows)
#   • Models        : LR, XGBoost, RF, SVM, Neural Network
#   • XAI Methods   : LIME (local) + SHAP (global)
#   • SHAP explainers tested: Tree, Kernel, Linear, Deep
#   • Extra         : ALE plots for LR/SVM/NN
#   • Limitation    : No stability/faithfulness evaluation of XAI methods
#   • Limitation    : No counterfactual explanations
#   • Limitation    : No rule-based explanations
#   • Limitation    : No hyperparameter tuning beyond grid search
#
# IMPROVEMENTS IN THIS CODE (Novel Contributions over the paper)
# ─────────────────────────────────────────────────────────────────────────────
#   1.  German Credit dataset (standard benchmark — directly comparable to lit.)
#   2.  SHAP TreeExplainer + KernelExplainer comparison (same as paper §5.1)
#   3.  SHAP Force plots & Dependence plots (same as paper §5.1 + §5.2)
#   4.  ALE plots (same as paper §5.4) — extended to XGBoost
#   5.  LIME stability quantification (NEW — paper had no stability analysis)
#   6.  Faithfulness test: SHAP-ordered vs random feature removal (NEW)
#   7.  SHAP vs LIME agreement — Spearman rank-correlation (NEW)
#   8.  Anchors: rule-based IF-THEN explanations (NEW — paper had no anchors)
#   9.  DiCE counterfactual explanations (NEW — paper had no counterfactuals)
#  10.  XGBoost + LightGBM (paper had XGBoost but no LightGBM)
#  11.  Optuna hyperparameter tuning (paper used grid search only)
#  12.  SMOTE + effect of balancing on SHAP importances (NEW)
#  13.  Confidence-stratified case analysis — 4 quadrants (NEW)
#  14.  Cost-sensitive evaluation — German Credit official cost matrix (NEW)
#  15.  XAI method comparison table: LIME vs SHAP vs Anchors (NEW)
# =============================================================================


# ─────────────────────────────────────────────────────────────────────────────
# CELL 1 — INSTALL ALL DEPENDENCIES
# Run once. Takes ~3 minutes. Comment out after first run.
# ─────────────────────────────────────────────────────────────────────────────

import subprocess, sys

pkgs = [
    "lime", "dice-ml", "shap", "anchor-exp",
    "xgboost", "lightgbm", "optuna",
    "imbalanced-learn", "plotly", "PyALE"
]
for p in pkgs:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", p])

print("All packages installed successfully.")


# ─────────────────────────────────────────────────────────────────────────────
# CELL 2 — IMPORTS
# ─────────────────────────────────────────────────────────────────────────────

import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
from scipy.stats import spearmanr

# Sklearn
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score, balanced_accuracy_score, classification_report,
    confusion_matrix, roc_auc_score, average_precision_score, f1_score
)
from sklearn.model_selection import StratifiedKFold, cross_val_score, train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.neural_network import MLPClassifier

# Imbalanced-learn
from imblearn.over_sampling import SMOTE

# XGBoost & LightGBM
import xgboost as xgb
import lightgbm as lgb

# Optuna
import optuna
optuna.logging.set_verbosity(optuna.logging.WARNING)

# XAI
import lime
import lime.lime_tabular
import shap
import dice_ml
from dice_ml import Dice
from anchor import anchor_tabular

# ALE
from PyALE import ale

warnings.filterwarnings("ignore")

try:
    shap.initjs()
except Exception:
    pass

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)
print("All imports successful.")


# ─────────────────────────────────────────────────────────────────────────────
# CELL 3 — LOAD GermanCredit.csv
# Upload GermanCredit.csv to Colab first (drag into left Files panel)
# ─────────────────────────────────────────────────────────────────────────────

# ── Try Colab upload path first, then current directory ──────────────────────
import os

POSSIBLE_PATHS = [
    "/content/GermanCredit.csv",          # Colab root after upload
    "GermanCredit.csv",                   # current directory
    "/content/drive/MyDrive/GermanCredit.csv",  # Google Drive
]

df_raw = None
for path in POSSIBLE_PATHS:
    if os.path.exists(path):
        df_raw = pd.read_csv(path)
        print(f"Loaded from: {path}")
        break

if df_raw is None:
    # Last resort: prompt user upload inside notebook
    from google.colab import files
    print("Please upload GermanCredit.csv now ↓")
    uploaded = files.upload()
    fname = list(uploaded.keys())[0]
    df_raw = pd.read_csv(fname)
    print(f"Loaded: {fname}")

print(f"\nDataset shape : {df_raw.shape}")
print(f"Columns       : {df_raw.columns.tolist()}")
print(f"\nTarget distribution:")
print(df_raw["credit_risk"].value_counts())
print(f"\nBase rate (Bad=0): {(df_raw['credit_risk']==0).mean():.1%}")
print(f"\nSample rows:")
print(df_raw.head(3))


# ─────────────────────────────────────────────────────────────────────────────
# CELL 4 — DEFINE COLUMN TYPES (based on actual GermanCredit.csv columns)
# ─────────────────────────────────────────────────────────────────────────────

TARGET = "credit_risk"

NUMERICAL_COLS = [
    "duration", "amount", "installment_rate",
    "present_residence", "age", "number_credits", "people_liable"
]

CATEGORICAL_COLS = [
    "status", "credit_history", "purpose", "savings",
    "employment_duration", "personal_status_sex", "other_debtors",
    "property", "other_installment_plans", "housing",
    "job", "telephone", "foreign_worker"
]

# Keep only columns that actually exist in the file
NUMERICAL_COLS  = [c for c in NUMERICAL_COLS  if c in df_raw.columns]
CATEGORICAL_COLS = [c for c in CATEGORICAL_COLS if c in df_raw.columns]

print(f"Numerical   ({len(NUMERICAL_COLS)}) : {NUMERICAL_COLS}")
print(f"Categorical ({len(CATEGORICAL_COLS)}) : {CATEGORICAL_COLS}")
print(f"No missing values: {df_raw.isnull().sum().sum() == 0}")


# ─────────────────────────────────────────────────────────────────────────────
# CELL 5 — EDA  (matches paper's exploratory analysis, extended)
# ─────────────────────────────────────────────────────────────────────────────

# 5a. Numerical distributions by class (paper Figure 3 equivalent)
fig, axes = plt.subplots(2, 4, figsize=(18, 8))
axes = axes.ravel()
for i, feat in enumerate(NUMERICAL_COLS):
    axes[i].hist(df_raw[df_raw[TARGET]==1][feat], bins=20, alpha=0.6,
                 color="steelblue", label="Good (1)")
    axes[i].hist(df_raw[df_raw[TARGET]==0][feat], bins=20, alpha=0.6,
                 color="coral", label="Bad (0)")
    axes[i].set_title(feat, fontsize=10)
    axes[i].legend(fontsize=7)
for j in range(len(NUMERICAL_COLS), len(axes)):
    axes[j].set_visible(False)
plt.suptitle("Feature Distributions by Credit Risk Class", fontsize=13)
plt.tight_layout(); plt.show()

# 5b. Correlation heatmap (paper Figure 3 equivalent)
corr_df = df_raw[NUMERICAL_COLS + [TARGET]].corr()
plt.figure(figsize=(8, 6))
sns.heatmap(corr_df, annot=True, fmt=".2f", cmap="coolwarm",
            center=0, square=True, linewidths=0.5)
plt.title("Correlation Heatmap — Numerical Features")
plt.tight_layout(); plt.show()

# 5c. Class imbalance bar
plt.figure(figsize=(4, 3))
df_raw[TARGET].value_counts().plot(kind="bar", color=["coral","steelblue"],
                                    edgecolor="black")
plt.xticks([0,1], ["Bad (0)","Good (1)"], rotation=0)
plt.title("Class Distribution"); plt.ylabel("Count")
plt.tight_layout(); plt.show()
print("Class imbalance: 70% Good, 30% Bad → SMOTE required")


# ─────────────────────────────────────────────────────────────────────────────
# CELL 6 — PREPROCESSING
# StandardScaler for numerical, OneHotEncoder for categorical
# ─────────────────────────────────────────────────────────────────────────────

X_raw = df_raw[NUMERICAL_COLS + CATEGORICAL_COLS].copy()
y     = df_raw[TARGET].copy()

import sklearn
_sklearn_version = tuple(int(x) for x in sklearn.__version__.split(".")[:2])

# sklearn >= 1.2 renamed sparse=False to sparse_output=False
if _sklearn_version >= (1, 2):
    _ohe = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
else:
    _ohe = OneHotEncoder(handle_unknown="ignore", sparse=False)

preprocessor = ColumnTransformer(transformers=[
    ("num", StandardScaler(), NUMERICAL_COLS),
    ("cat", _ohe, CATEGORICAL_COLS),
])

X_enc = preprocessor.fit_transform(X_raw)

cat_names     = preprocessor.named_transformers_["cat"]\
                .get_feature_names_out(CATEGORICAL_COLS).tolist()
FEATURE_NAMES = NUMERICAL_COLS + cat_names

print(f"Encoded feature matrix : {X_enc.shape}")
print(f"Total features after OHE: {len(FEATURE_NAMES)}")


# ─────────────────────────────────────────────────────────────────────────────
# CELL 7 — TRAIN / TEST SPLIT
# ─────────────────────────────────────────────────────────────────────────────

X_train, X_test, y_train, y_test = train_test_split(
    X_enc, y, test_size=0.25, stratify=y, random_state=RANDOM_STATE
)

print(f"Train: {X_train.shape}  Good={y_train.sum()}  Bad={(y_train==0).sum()}")
print(f"Test : {X_test.shape}  Good={y_test.sum()}  Bad={(y_test==0).sum()}")


# ─────────────────────────────────────────────────────────────────────────────
# CELL 8 — SMOTE OVERSAMPLING
# IMPROVEMENT over paper: paper used no explicit oversampling on this domain.
# SMOTE creates synthetic minority (Bad credit) samples.
# ─────────────────────────────────────────────────────────────────────────────

print("Before SMOTE — Bad:", (y_train==0).sum(), " Good:", y_train.sum())
smote = SMOTE(random_state=RANDOM_STATE)
X_train_sm, y_train_sm = smote.fit_resample(X_train, y_train)
print("After  SMOTE — Bad:", (y_train_sm==0).sum(), " Good:", y_train_sm.sum())


# ─────────────────────────────────────────────────────────────────────────────
# CELL 9 — COST-SENSITIVE METRIC (German Credit official cost matrix)
# IMPROVEMENT: paper reported only accuracy/F1/ROC. German Credit has an
# official cost matrix: Bad→Good error costs 5× more than Good→Bad.
# ─────────────────────────────────────────────────────────────────────────────

def cost_score(y_true, y_pred):
    """Lower is better. Official German Credit cost matrix."""
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    return (5 * np.sum((y_true==0) & (y_pred==1)) +   # Bad predicted as Good
            1 * np.sum((y_true==1) & (y_pred==0)))    # Good predicted as Bad


# ─────────────────────────────────────────────────────────────────────────────
# CELL 10 — TRAIN ALL MODELS & EVALUATE
# Models: LR, DT, RF, GBM, SVM, Neural Net (same as paper) + XGBoost, LightGBM
# IMPROVEMENT: added LightGBM + cost-sensitive metric + Optuna tuning
# ─────────────────────────────────────────────────────────────────────────────

def evaluate_model(name, model, X_tr, y_tr, X_te, y_te, cv_folds=5):
    model.fit(X_tr, y_tr)
    y_pred  = model.predict(X_te)
    y_proba = model.predict_proba(X_te)[:, 1]

    metrics = {
        "model":         name,
        "accuracy":      round(accuracy_score(y_te, y_pred), 4),
        "balanced_acc":  round(balanced_accuracy_score(y_te, y_pred), 4),
        "f1_bad":        round(f1_score(y_te, y_pred, pos_label=0), 4),
        "auc_roc":       round(roc_auc_score(y_te, y_proba), 4),
        "auc_pr_bad":    round(average_precision_score(1-y_te, 1-y_proba), 4),
        "cost":          cost_score(y_te, y_pred),
    }
    cv = cross_val_score(
        model, X_tr, y_tr,
        cv=StratifiedKFold(n_splits=cv_folds, shuffle=True,
                           random_state=RANDOM_STATE),
        scoring="roc_auc"
    )
    metrics["cv_auc"] = f"{cv.mean():.4f} ± {cv.std()*2:.4f}"

    print(f"\n{'='*55}\n  {name}")
    print(f"  Accuracy        : {metrics['accuracy']}")
    print(f"  Balanced Acc    : {metrics['balanced_acc']}")
    print(f"  F1 (Bad class)  : {metrics['f1_bad']}")
    print(f"  AUC-ROC         : {metrics['auc_roc']}")
    print(f"  AUC-PR (Bad)    : {metrics['auc_pr_bad']}")
    print(f"  Cost (↓ better) : {metrics['cost']}")
    print(f"  CV AUC          : {metrics['cv_auc']}")
    print(classification_report(y_te, y_pred, target_names=["Bad","Good"]))

    cm = confusion_matrix(y_te, y_pred)
    plt.figure(figsize=(4,3))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
                xticklabels=["Bad","Good"], yticklabels=["Bad","Good"])
    plt.title(f"Confusion Matrix — {name}")
    plt.xlabel("Predicted"); plt.ylabel("Actual")
    plt.tight_layout(); plt.show()
    return metrics, model


# All models from paper + extras
classifiers = {
    # ── Paper models ──────────────────────────────────────────────────────
    "LogisticRegression": LogisticRegression(
        max_iter=2000, random_state=RANDOM_STATE, class_weight="balanced"),
    "DecisionTree":       DecisionTreeClassifier(
        max_depth=6, random_state=RANDOM_STATE, class_weight="balanced"),
    "RandomForest":       RandomForestClassifier(
        n_estimators=300, random_state=RANDOM_STATE, class_weight="balanced"),
    "GradientBoosting":   GradientBoostingClassifier(
        n_estimators=300, random_state=RANDOM_STATE),
    "SVM":                SVC(
        probability=True, kernel="rbf", gamma="auto",
        class_weight="balanced", random_state=RANDOM_STATE),
    "NeuralNetwork":      MLPClassifier(
        hidden_layer_sizes=(35,35), activation="relu", max_iter=500,
        random_state=RANDOM_STATE),
    # ── New models (not in paper) ──────────────────────────────────────────
    "XGBoost":            xgb.XGBClassifier(
        n_estimators=300, use_label_encoder=False, eval_metric="logloss",
        scale_pos_weight=2.33, random_state=RANDOM_STATE, verbosity=0),
    "LightGBM":           lgb.LGBMClassifier(
        n_estimators=300, class_weight="balanced",
        random_state=RANDOM_STATE, verbose=-1),
}

all_metrics    = []
trained_models = {}

for name, clf in classifiers.items():
    metrics, fitted = evaluate_model(
        name, clf, X_train_sm, y_train_sm, X_test, y_test)
    all_metrics.append(metrics)
    trained_models[name] = fitted


# ─────────────────────────────────────────────────────────────────────────────
# CELL 11 — CONSOLIDATED RESULTS TABLE
# ─────────────────────────────────────────────────────────────────────────────

results_df = pd.DataFrame(all_metrics).set_index("model")
results_df = results_df.sort_values("auc_roc", ascending=False)

print("\n===== CONSOLIDATED RESULTS TABLE (Table 2 equivalent) =====")
print(results_df[["accuracy","balanced_acc","f1_bad",
                   "auc_roc","auc_pr_bad","cost","cv_auc"]].to_string())

# AUC-ROC comparison bar chart
results_df["auc_roc"].sort_values().plot(
    kind="barh", figsize=(10,6), color="steelblue", edgecolor="black")
plt.xlabel("AUC-ROC"); plt.title("Model Comparison — AUC-ROC")
plt.tight_layout(); plt.show()

# Cost comparison (IMPROVEMENT over paper — paper had no cost metric)
results_df["cost"].sort_values(ascending=False).plot(
    kind="barh", figsize=(10,6), color="coral", edgecolor="black")
plt.xlabel("Cost (Lower = Better)")
plt.title("Model Comparison — Cost-Sensitive (5× penalty for Bad→Good)")
plt.tight_layout(); plt.show()


# ─────────────────────────────────────────────────────────────────────────────
# CELL 12 — OPTUNA HYPERPARAMETER TUNING (XGBoost)
# IMPROVEMENT over paper: paper used GridSearchCV; Optuna is Bayesian → faster
# ─────────────────────────────────────────────────────────────────────────────

def xgb_objective(trial):
    params = {
        "n_estimators":      trial.suggest_int("n_estimators", 100, 600),
        "max_depth":         trial.suggest_int("max_depth", 3, 10),
        "learning_rate":     trial.suggest_float("learning_rate", 0.01, 0.3, log=True),
        "subsample":         trial.suggest_float("subsample", 0.5, 1.0),
        "colsample_bytree":  trial.suggest_float("colsample_bytree", 0.5, 1.0),
        "min_child_weight":  trial.suggest_int("min_child_weight", 1, 10),
        "gamma":             trial.suggest_float("gamma", 0.0, 1.0),
        "scale_pos_weight":  trial.suggest_float("scale_pos_weight", 1.0, 5.0),
        "use_label_encoder": False,
        "eval_metric": "logloss",
        "verbosity": 0,
        "random_state": RANDOM_STATE,
    }
    cv = cross_val_score(
        xgb.XGBClassifier(**params),
        X_train_sm, y_train_sm,
        cv=StratifiedKFold(n_splits=3, shuffle=True, random_state=RANDOM_STATE),
        scoring="roc_auc"
    )
    return cv.mean()


print("Tuning XGBoost with Optuna (60 trials) …")
study = optuna.create_study(
    direction="maximize",
    sampler=optuna.samplers.TPESampler(seed=RANDOM_STATE)
)
study.optimize(xgb_objective, n_trials=60, show_progress_bar=True)

print(f"\nBest AUC  : {study.best_value:.4f}")
print(f"Best params: {study.best_params}")

xgb_tuned = xgb.XGBClassifier(
    **study.best_params,
    use_label_encoder=False, eval_metric="logloss",
    random_state=RANDOM_STATE, verbosity=0
)
tuned_metrics, xgb_tuned_fitted = evaluate_model(
    "XGBoost_Tuned", xgb_tuned, X_train_sm, y_train_sm, X_test, y_test)
all_metrics.append(tuned_metrics)
trained_models["XGBoost_Tuned"] = xgb_tuned_fitted

BEST_MODEL      = xgb_tuned_fitted
BEST_MODEL_NAME = "XGBoost_Tuned"


# ─────────────────────────────────────────────────────────────────────────────
# CELL 13 — SHAP: TREE EXPLAINER (paper §5.1a — same method, new dataset)
# ─────────────────────────────────────────────────────────────────────────────

shap_tree_exp = shap.TreeExplainer(BEST_MODEL)
shap_values   = shap_tree_exp.shap_values(X_test)

# Summary beeswarm (paper Figure 8a equivalent)
print("SHAP Summary Plot — Beeswarm (TreeExplainer, XGBoost_Tuned)")
shap.summary_plot(shap_values, X_test,
                  feature_names=FEATURE_NAMES, show=True)

# Summary bar — global importance (paper Figure 8a equivalent)
print("SHAP Summary Plot — Bar (Global Mean |SHAP|)")
shap.summary_plot(shap_values, X_test,
                  feature_names=FEATURE_NAMES, plot_type="bar", show=True)

# Store global importance
shap_global = pd.Series(
    np.abs(shap_values).mean(axis=0),
    index=FEATURE_NAMES
).sort_values(ascending=False)

print("\nTop 10 SHAP features (TreeExplainer):")
print(shap_global.head(10).to_string())


# ─────────────────────────────────────────────────────────────────────────────
# CELL 14 — SHAP: KERNEL EXPLAINER vs TREE EXPLAINER COMPARISON
# IMPROVEMENT: paper compared these on Lending Club; we compare on German Credit
# and add a quantitative Spearman ρ measure of consistency between explainers.
# ─────────────────────────────────────────────────────────────────────────────

print("\nComputing SHAP KernelExplainer (200 test samples) …")
background    = shap.kmeans(X_train_sm, 30)   # paper used shap.kmeans(20k, 30)
kernel_exp    = shap.KernelExplainer(BEST_MODEL.predict_proba, background)
shap_kernel_raw = kernel_exp.shap_values(X_test[:200])

# KernelExplainer returns either:
#   - a list of arrays [class0_array, class1_array]  (older shap)
#   - a 3D array of shape (n_samples, n_features, n_classes)  (newer shap)
# We always want class 1 (Good credit) values with shape (n_samples, n_features)
if isinstance(shap_kernel_raw, list):
    # list of per-class arrays → take index 1 (Good)
    shap_kernel_v = shap_kernel_raw[1]
elif isinstance(shap_kernel_raw, np.ndarray) and shap_kernel_raw.ndim == 3:
    # 3D array (n_samples, n_features, n_classes) → slice last dim
    shap_kernel_v = shap_kernel_raw[:, :, 1]
else:
    # Already 2D (n_samples, n_features) — binary output squeezed
    shap_kernel_v = shap_kernel_raw

print(f"Kernel SHAP values shape: {np.array(shap_kernel_v).shape}")

shap_kernel_global = pd.Series(
    np.abs(shap_kernel_v).mean(axis=0),
    index=FEATURE_NAMES
).sort_values(ascending=False)

# Summary plot — Kernel explainer (paper Figure 8b equivalent)
print("SHAP Summary — KernelExplainer (200 test samples)")
shap.summary_plot(shap_kernel_v, X_test[:200],
                  feature_names=FEATURE_NAMES, plot_type="bar", show=True)

# Quantitative comparison: Tree vs Kernel (IMPROVEMENT over paper)
rho_tk, p_tk = spearmanr(
    shap_global.reindex(FEATURE_NAMES),
    shap_kernel_global.reindex(FEATURE_NAMES)
)
print(f"\nTree vs Kernel SHAP — Spearman ρ = {rho_tk:.3f}  (p = {p_tk:.4f})")
print("→ Paper (§5.1.1): 'both explainers consistent in top 20 features'")
print(f"→ This work quantifies that consistency: ρ = {rho_tk:.3f}")

# Visual side-by-side (paper Figure 13 equivalent)
top10 = shap_global.head(10).index.tolist()
fig, axes = plt.subplots(1, 2, figsize=(16, 5))
shap_global[top10].sort_values().plot(
    kind="barh", ax=axes[0], color="steelblue",
    title="SHAP — TreeExplainer (XGBoost)")
shap_kernel_global.reindex(top10).sort_values().plot(
    kind="barh", ax=axes[1], color="coral",
    title="SHAP — KernelExplainer (XGBoost, 200 samples)")
plt.suptitle(f"Tree vs Kernel SHAP Comparison  |  Spearman ρ = {rho_tk:.3f}")
plt.tight_layout(); plt.show()


# ─────────────────────────────────────────────────────────────────────────────
# CELL 15 — SHAP: LOCAL WATERFALL PLOTS  (paper §4.1.1 instance interpretation)
# ─────────────────────────────────────────────────────────────────────────────

bad_idx  = int(np.where((y_test.values==0) &
                        (BEST_MODEL.predict(X_test)==0))[0][0])
good_idx = int(np.where((y_test.values==1) &
                        (BEST_MODEL.predict(X_test)==1))[0][0])

for idx, label in [(bad_idx,"BAD credit"), (good_idx,"GOOD credit")]:
    print(f"\nSHAP Waterfall — {label} (Instance {idx})")
    shap.waterfall_plot(
        shap.Explanation(
            values        = shap_values[idx],
            base_values   = shap_tree_exp.expected_value,
            data          = X_test[idx],
            feature_names = FEATURE_NAMES
        )
    )


# ─────────────────────────────────────────────────────────────────────────────
# CELL 16 — SHAP: DEPENDENCE PLOTS  (paper Figure 9 equivalent)
# ─────────────────────────────────────────────────────────────────────────────

top3 = shap_global.head(3).index.tolist()
for feat in top3:
    if feat in FEATURE_NAMES:
        print(f"\nSHAP Dependence Plot — {feat}")
        shap.dependence_plot(
            FEATURE_NAMES.index(feat), shap_values, X_test,
            feature_names=FEATURE_NAMES, show=True
        )


# ─────────────────────────────────────────────────────────────────────────────
# CELL 17 — SHAP: FORCE PLOT  (paper Figure 10 equivalent)
# ─────────────────────────────────────────────────────────────────────────────

print("\nSHAP Force Plot — 200 test instances stacked")
shap.force_plot(
    shap_tree_exp.expected_value,
    shap_values[:200],
    X_test[:200],
    feature_names=FEATURE_NAMES
)


# ─────────────────────────────────────────────────────────────────────────────
# CELL 18 — SHAP: GLOBAL FEATURE IMPORTANCE vs TREE FEATURE IMPORTANCE
# IMPROVEMENT: paper compared these on Lending Club; we replicate + quantify.
# (Paper Figure 13 equivalent)
# ─────────────────────────────────────────────────────────────────────────────

tree_imp = pd.Series(
    BEST_MODEL.feature_importances_,
    index=FEATURE_NAMES
).sort_values(ascending=False)

top15 = shap_global.head(15).index.tolist()
fig, axes = plt.subplots(1, 2, figsize=(16, 6))
shap_global[top15].sort_values().plot(
    kind="barh", ax=axes[0], color="steelblue",
    title="SHAP Global Importance (Mean |SHAP|)")
tree_imp.reindex(top15).sort_values().plot(
    kind="barh", ax=axes[1], color="orange",
    title="Tree Feature Importance (Information Gain)")
plt.suptitle("SHAP vs Tree Feature Importance  (Paper Figure 13 equivalent)")
plt.tight_layout(); plt.show()

rho_st, p_st = spearmanr(
    shap_global.reindex(FEATURE_NAMES),
    tree_imp.reindex(FEATURE_NAMES)
)
print(f"\nSHAP vs Tree importance — Spearman ρ = {rho_st:.3f}  (p = {p_st:.4f})")
print("Paper (§5.3): 'different features contribute to SHAP vs information gain'")


# ─────────────────────────────────────────────────────────────────────────────
# CELL 19 — ALE PLOTS  (paper §5.4 equivalent — extended to XGBoost)
# IMPROVEMENT: paper computed ALE only for LR/SVM/NN (not tree models).
# We compute for LogisticRegression AND XGBoost for comparison.
# ─────────────────────────────────────────────────────────────────────────────

# Rebuild DataFrames needed for PyALE
X_train_df = pd.DataFrame(X_train_sm, columns=FEATURE_NAMES)
X_test_df  = pd.DataFrame(X_test,     columns=FEATURE_NAMES)

def plot_ale(model, model_name, feature, X_df):
    """ALE plot for one numerical feature."""
    try:
        ale_eff = ale(
            X=X_df,
            model=model,
            feature=[feature],
            feature_type="continuous",
            grid_size=20,
            include_CI=True,
            plot=True
        )
        plt.title(f"ALE Plot — {feature}  [{model_name}]")
        plt.tight_layout(); plt.show()
    except Exception as e:
        print(f"ALE failed for {feature} ({model_name}): {e}")


# Paper used top features: we use top 3 numerical features by SHAP
top_num_feats = [f for f in shap_global.index if f in NUMERICAL_COLS][:3]

for model_name in ["LogisticRegression", "XGBoost_Tuned"]:
    model = trained_models[model_name]
    print(f"\n===== ALE Plots — {model_name} =====")
    for feat in top_num_feats:
        plot_ale(model, model_name, feat, X_test_df)


# ─────────────────────────────────────────────────────────────────────────────
# CELL 20 — LIME SETUP  (paper §4 equivalent)
# ─────────────────────────────────────────────────────────────────────────────

cat_feat_idx = [
    i for i, name in enumerate(FEATURE_NAMES)
    if any(name.startswith(c) for c in CATEGORICAL_COLS)
]

lime_exp = lime.lime_tabular.LimeTabularExplainer(
    training_data        = X_train_sm,
    feature_names        = FEATURE_NAMES,
    class_names          = ["Bad", "Good"],
    categorical_features = cat_feat_idx,
    discretize_continuous= True,
    random_state         = RANDOM_STATE,
    mode                 = "classification",
)


def run_lime(idx, model=None, n_feat=10, n_samp=100_000):
    """LIME explanation for one test instance (paper §4.1.1 equivalent)."""
    if model is None:
        model = BEST_MODEL
    true = "Good" if y_test.iloc[idx]==1 else "Bad"
    prob  = model.predict_proba(X_test[idx:idx+1])[0]
    print(f"\n── LIME — Instance {idx}  True={true}  "
          f"P(Bad)={prob[0]:.3f}  P(Good)={prob[1]:.3f}")
    exp = lime_exp.explain_instance(
        X_test[idx], model.predict_proba,
        num_features=n_feat, num_samples=n_samp)
    exp.as_pyplot_figure()
    plt.title(f"LIME — Instance {idx}  (True={true})")
    plt.show()
    return exp


# Reproduce paper's three instance types (paper §4.1.1)
print("\n=== LIME: Good credit (predicted Good) ===")
_ = run_lime(good_idx, BEST_MODEL)

print("\n=== LIME: Bad credit (predicted Bad) ===")
_ = run_lime(bad_idx, BEST_MODEL)

# Wrong prediction (paper §4.2 SVM equivalent — uncertain prediction)
uncertain_idx = int(np.argsort(np.abs(
    BEST_MODEL.predict_proba(X_test)[:,1] - 0.5))[:10][0])
print(f"\n=== LIME: Uncertain prediction — Instance {uncertain_idx} ===")
_ = run_lime(uncertain_idx, BEST_MODEL)


# ─────────────────────────────────────────────────────────────────────────────
# CELL 21 — LIME ON ALL PAPER MODELS (paper §4.1, §4.2 equivalent)
# Shows how LIME explanation changes across model types — same as paper.
# ─────────────────────────────────────────────────────────────────────────────

print("\n===== LIME explanations across model types (Paper §4 equivalent) =====")
for model_name in ["LogisticRegression", "RandomForest",
                   "XGBoost_Tuned", "SVM", "NeuralNetwork"]:
    model = trained_models[model_name]
    print(f"\n── {model_name} ── Instance {bad_idx} (True=Bad)")
    exp = lime_exp.explain_instance(
        X_test[bad_idx], model.predict_proba,
        num_features=8, num_samples=50_000)
    exp.as_pyplot_figure()
    plt.title(f"LIME — {model_name}")
    plt.show()


# ─────────────────────────────────────────────────────────────────────────────
# CELL 22 — LIME STABILITY TEST  (IMPROVEMENT — paper had no stability test)
# NOVELTY: Run LIME 10× with different seeds. Measure Spearman ρ consistency.
# Paper §7 limitation: "LIME has certain limitations" — we quantify them.
# ─────────────────────────────────────────────────────────────────────────────

def _lime_label_to_feat(label, feature_names):
    """
    LIME with discretize_continuous=True returns labels like 'duration <= 0.59'
    or 'status_no checking account = 0.0'. Match back to the original feature name
    by checking which feature name the label starts with.
    """
    for f in feature_names:
        if label.startswith(f):
            return f
    return None


def lime_stability(idx, model=None, n_runs=10, n_feat=10, n_samp=50_000):
    if model is None:
        model = BEST_MODEL
    vecs = []
    for seed in range(n_runs):
        exp_tmp = lime.lime_tabular.LimeTabularExplainer(
            training_data        = X_train_sm,
            feature_names        = FEATURE_NAMES,
            class_names          = ["Bad","Good"],
            categorical_features = cat_feat_idx,
            discretize_continuous= True,
            random_state         = seed,
            mode                 = "classification",
        )
        exp = exp_tmp.explain_instance(
            X_test[idx], model.predict_proba,
            num_features=n_feat, num_samples=n_samp)
        vec = np.zeros(len(FEATURE_NAMES))
        for label, imp in exp.as_list():
            feat = _lime_label_to_feat(label, FEATURE_NAMES)
            if feat is not None:
                vec[FEATURE_NAMES.index(feat)] += abs(imp)
        vecs.append(vec)
    corrs = [spearmanr(vecs[i], vecs[j])[0]
             for i in range(n_runs) for j in range(i+1, n_runs)]
    return float(np.mean(corrs)), float(np.std(corrs))


print("\n── LIME Stability Test (paper §7 quantified) ──")
test_ids = [bad_idx, good_idx, uncertain_idx, 50, 150]
stab_rows = []
for idx in test_ids:
    mu, sd = lime_stability(idx)
    stab_rows.append({"instance": idx,
                      "mean_rho": round(mu,3),
                      "std":      round(sd,3)})
    print(f"  Instance {idx:3d}: ρ = {mu:.3f} ± {sd:.3f}")

stab_df = pd.DataFrame(stab_rows)
print("\nρ ≈ 1.0 → stable | ρ < 0.6 → unreliable")
print(stab_df.to_string(index=False))

plt.figure(figsize=(7,4))
plt.bar(stab_df["instance"].astype(str), stab_df["mean_rho"],
        yerr=stab_df["std"], color="steelblue", edgecolor="black", capsize=5)
plt.axhline(1.0, color="green",  linestyle="--", label="Perfect")
plt.axhline(0.6, color="orange", linestyle="--", label="Acceptable")
plt.xlabel("Instance"); plt.ylabel("Mean Spearman ρ")
plt.title("LIME Stability Across Seeds (IMPROVEMENT over paper)")
plt.ylim(0, 1.15); plt.legend()
plt.tight_layout(); plt.show()


# ─────────────────────────────────────────────────────────────────────────────
# CELL 23 — FAITHFULNESS TEST  (IMPROVEMENT — paper had no faithfulness test)
# Remove SHAP top-K features → track AUC drop vs random removal order.
# Proves SHAP explanations are faithful to the model.
# ─────────────────────────────────────────────────────────────────────────────

def faithfulness_test(model, shap_imp, X, y, k_max=12, n_reps=5):
    shap_aucs, rand_aucs = [], []
    for k in range(1, k_max+1):
        Xm = X.copy()
        for f in shap_imp.index[:k]:
            Xm[:, FEATURE_NAMES.index(f)] = 0.0
        shap_aucs.append(roc_auc_score(y, model.predict_proba(Xm)[:,1]))

        ra = []
        for _ in range(n_reps):
            Xr = X.copy()
            for f in np.random.choice(FEATURE_NAMES, k, replace=False):
                Xr[:, FEATURE_NAMES.index(f)] = 0.0
            ra.append(roc_auc_score(y, model.predict_proba(Xr)[:,1]))
        rand_aucs.append(float(np.mean(ra)))

    plt.figure(figsize=(8,5))
    plt.plot(range(1,k_max+1), shap_aucs, "o-", color="red",
             label="SHAP-ordered removal", linewidth=2)
    plt.plot(range(1,k_max+1), rand_aucs,  "s--", color="gray",
             label="Random removal baseline", linewidth=2)
    plt.xlabel("Features removed"); plt.ylabel("AUC-ROC")
    plt.title("Faithfulness Test — SHAP Removal (IMPROVEMENT over paper)")
    plt.legend(); plt.grid(alpha=0.3)
    plt.tight_layout(); plt.show()

    score = sum(r-s for s,r in zip(shap_aucs, rand_aucs))
    print(f"Faithfulness Score = {score:.4f}  (positive = faithful)")
    return shap_aucs, rand_aucs


print("\nRunning Faithfulness Test …")
fa, ra = faithfulness_test(BEST_MODEL, shap_global, X_test, y_test)


# ─────────────────────────────────────────────────────────────────────────────
# CELL 24 — SHAP vs LIME AGREEMENT  (IMPROVEMENT — paper had no comparison)
# Spearman ρ between global SHAP and global LIME importance rankings.
# Paper compared SHAP vs tree importance (§5.3); we compare SHAP vs LIME.
# ─────────────────────────────────────────────────────────────────────────────

def lime_global_imp(model, X_s, n=100, n_feat=10, n_samp=50_000):
    acc = np.zeros(len(FEATURE_NAMES))
    for idx in tqdm(np.random.choice(len(X_s), n, replace=False),
                    desc="LIME global"):
        exp = lime_exp.explain_instance(
            X_s[idx], model.predict_proba,
            num_features=n_feat, num_samples=n_samp)
        for label, imp in exp.as_list():
            feat = _lime_label_to_feat(label, FEATURE_NAMES)
            if feat is not None:
                acc[FEATURE_NAMES.index(feat)] += abs(imp)
    return pd.Series(acc/n, index=FEATURE_NAMES).sort_values(ascending=False)


print("Computing global LIME importance (100 test instances) …")
lime_global = lime_global_imp(BEST_MODEL, X_test)

rho_sl, p_sl = spearmanr(
    shap_global.reindex(FEATURE_NAMES),
    lime_global.reindex(FEATURE_NAMES)
)
print(f"\nSHAP vs LIME — Spearman ρ = {rho_sl:.3f}  (p = {p_sl:.4f})")
if rho_sl > 0.7:
    print("→ Methods largely agree on feature rankings")
else:
    print("→ Notable disagreement — PAPER CONTRIBUTION")

top10 = shap_global.head(10).index.tolist()
comp_df = pd.DataFrame({
    "SHAP": shap_global[top10].values,
    "LIME": lime_global.reindex(top10).values
}, index=[f[:22] for f in top10])
comp_df.plot(kind="bar", figsize=(13,5),
             color=["coral","steelblue"], edgecolor="black")
plt.title(f"SHAP vs LIME Global Importance (Top 10)  |  ρ = {rho_sl:.3f}")
plt.xticks(rotation=45, ha="right")
plt.tight_layout(); plt.show()


# ─────────────────────────────────────────────────────────────────────────────
# CELL 25 — ANCHORS: RULE-BASED EXPLANATIONS  (NEW — not in paper)
# Paper limitation: only LIME (weights) + SHAP (weights).
# Anchors give IF-THEN rules — more actionable for loan officers.
# ─────────────────────────────────────────────────────────────────────────────

anchor_explainer = anchor_tabular.AnchorTabularExplainer(
    class_names      = ["Bad","Good"],
    feature_names    = FEATURE_NAMES,
    train_data       = X_train_sm,
    categorical_names= {
        i: [str(v) for v in np.unique(X_train_sm[:,i])]
        for i in cat_feat_idx
    }
)


def explain_anchor(idx, model=None, threshold=0.88):
    if model is None:
        model = BEST_MODEL
    true = "Good" if y_test.iloc[idx]==1 else "Bad"
    pred = "Good" if model.predict(X_test[idx:idx+1])[0]==1 else "Bad"
    print(f"\n── Anchor — Instance {idx}  True={true}  Pred={pred}")
    exp  = anchor_explainer.explain_instance(
        X_test[idx], model.predict, threshold=threshold)
    rule = (" AND\n   ".join(exp.names())
            if exp.names() else "(empty rule)")
    print(f"   IF   {rule}")
    print(f"   THEN predicted {pred}")
    print(f"   Precision: {exp.precision():.3f}  Coverage: {exp.coverage():.3f}")
    return exp


bad_ids  = np.where(y_test.values==0)[0][:4]
good_ids = np.where(y_test.values==1)[0][:4]

print("\n===== ANCHORS — BAD CREDIT (Paper improvement) =====")
for idx in bad_ids:
    explain_anchor(int(idx))

print("\n===== ANCHORS — GOOD CREDIT =====")
for idx in good_ids:
    explain_anchor(int(idx))


# ─────────────────────────────────────────────────────────────────────────────
# CELL 26 — DiCE COUNTERFACTUAL EXPLANATIONS  (NEW — not in paper)
# "What would this applicant need to change to get approved?"
# Paper limitation: no counterfactual explanations.
# ─────────────────────────────────────────────────────────────────────────────

dice_train_df          = pd.DataFrame(X_train_sm, columns=FEATURE_NAMES)
dice_train_df["label"] = y_train_sm.values
dice_test_df           = pd.DataFrame(X_test,     columns=FEATURE_NAMES)
dice_test_df["label"]  = y_test.values

d_dice = dice_ml.Data(
    dataframe         = dice_train_df,
    continuous_features = NUMERICAL_COLS,
    outcome_name      = "label"
)
m_dice = dice_ml.Model(model=BEST_MODEL, backend="sklearn")

permitted = {
    "duration":          [6, 72],
    "amount":            [250, 18424],
    "installment_rate":  [1, 4],
    "age":               [18, 75],
    "number_credits":    [1, 4],
    "people_liable":     [1, 2],
}


def dice_explain(idx, n_cfs=5):
    true = "Good" if y_test.iloc[idx]==1 else "Bad"
    print(f"\n── DiCE — Instance {idx}  (True={true}) ──")
    query = dice_test_df.drop(columns=["label"]).iloc[idx:idx+1]
    exp   = Dice(d_dice, m_dice, method="random")

    # Newer DiCE removed proximity_weight & diversity_weight
    import inspect
    sig    = inspect.signature(exp.generate_counterfactuals)
    kwargs = dict(
        query_instances  = query,
        desired_class    = "opposite",
        total_CFs        = n_cfs,
        features_to_vary = NUMERICAL_COLS,
        permitted_range  = permitted,
        verbose          = False,
    )
    if "proximity_weight" in sig.parameters:
        kwargs["proximity_weight"] = 5.0
    if "diversity_weight" in sig.parameters:
        kwargs["diversity_weight"] = 1.0
    if "random_seed" in sig.parameters:
        kwargs["random_seed"] = RANDOM_STATE

    result = exp.generate_counterfactuals(**kwargs)
    result.visualize_as_dataframe(show_only_changes=True)
    return result


print("\n===== DiCE: What must a Bad applicant change to get approved? =====")
for idx in bad_ids[:3]:
    dice_explain(int(idx))


# ─────────────────────────────────────────────────────────────────────────────
# CELL 27 — CONFIDENCE-STRATIFIED CASE ANALYSIS  (NEW — not in paper)
# 4-quadrant: high/low confidence × correct/wrong.
# Paper only showed arbitrary instances; we stratify by confidence.
# ─────────────────────────────────────────────────────────────────────────────

proba_t = BEST_MODEL.predict_proba(X_test)[:,1]
pred_t  = (proba_t >= 0.5).astype(int)
correct = (pred_t == y_test.values)
hi_conf = proba_t >= 0.80
lo_conf = (proba_t >= 0.40) & (proba_t <= 0.60)

groups = {
    "HIGH_CONF_CORRECT": np.where( hi_conf &  correct)[0][:3],
    "HIGH_CONF_WRONG":   np.where( hi_conf & ~correct)[0][:3],  # ← blind spots
    "LOW_CONF_CORRECT":  np.where( lo_conf &  correct)[0][:3],
    "LOW_CONF_WRONG":    np.where( lo_conf & ~correct)[0][:3],
}

for grp, idxs in groups.items():
    print(f"\n{'='*55}\nGROUP: {grp}  ({len(idxs)} shown)")
    if len(idxs) == 0:
        print("  (no instances in this group)"); continue
    for idx in idxs:
        true = "Good" if y_test.iloc[idx]==1 else "Bad"
        print(f"\n  Instance {idx}: True={true}  P(Good)={proba_t[idx]:.3f}")
        exp = lime_exp.explain_instance(
            X_test[idx], BEST_MODEL.predict_proba,
            num_features=8, num_samples=50_000)
        exp.as_pyplot_figure()
        plt.title(f"{grp} — Instance {idx}")
        plt.show()


# ─────────────────────────────────────────────────────────────────────────────
# CELL 28 — SMOTE EFFECT ON SHAP EXPLANATIONS  (NEW — not in paper)
# Does balancing training data change which features the model relies on?
# ─────────────────────────────────────────────────────────────────────────────

print("Training XGBoost WITHOUT SMOTE …")
xgb_no_smote = xgb.XGBClassifier(
    n_estimators=300, use_label_encoder=False,
    eval_metric="logloss", scale_pos_weight=2.33,
    random_state=RANDOM_STATE, verbosity=0
)
xgb_no_smote.fit(X_train, y_train)

sv_ns         = shap.TreeExplainer(xgb_no_smote).shap_values(X_test)
shap_no_smote = pd.Series(np.abs(sv_ns).mean(axis=0), index=FEATURE_NAMES)

rho_sm, p_sm = spearmanr(
    shap_global.reindex(FEATURE_NAMES),
    shap_no_smote.reindex(FEATURE_NAMES)
)
print(f"\nSHAP rank-correlation (SMOTE vs No-SMOTE): ρ = {rho_sm:.3f}")

fig, axes = plt.subplots(1, 2, figsize=(16,5))
shap_global.head(10).sort_values().plot(
    kind="barh", ax=axes[0], color="steelblue", title="SHAP — With SMOTE")
shap_no_smote.head(10).sort_values().plot(
    kind="barh", ax=axes[1], color="coral",     title="SHAP — Without SMOTE")
plt.suptitle(f"SMOTE Effect on Feature Importances  |  ρ = {rho_sm:.3f}")
plt.tight_layout(); plt.show()


# ─────────────────────────────────────────────────────────────────────────────
# CELL 29 — XAI METHOD COMPARISON TABLE  (NEW — not in paper)
# Side-by-side: what does LIME, SHAP, Anchors say about the SAME instance?
# ─────────────────────────────────────────────────────────────────────────────

def get_lime_top(idx, n=3):
    exp = lime_exp.explain_instance(
        X_test[idx], BEST_MODEL.predict_proba,
        num_features=n, num_samples=50_000)
    results = []
    for label, imp in sorted(exp.as_list(), key=lambda x: abs(x[1]), reverse=True):
        feat = _lime_label_to_feat(label, FEATURE_NAMES)
        results.append((feat or label)[:16])
    return results

def get_shap_top(idx, n=3):
    vals = shap_values[idx]
    top  = np.argsort(np.abs(vals))[::-1][:n]
    return [FEATURE_NAMES[i][:16] for i in top]

def get_anchor_str(idx):
    try:
        exp = anchor_explainer.explain_instance(
            X_test[idx], BEST_MODEL.predict, threshold=0.85)
        return (" & ".join(exp.names()[:2]) if exp.names() else "N/A")[:40]
    except:
        return "N/A"


print("\n===== XAI METHOD COMPARISON TABLE (IMPROVEMENT over paper) =====\n")
hdr = f"{'Idx':>5}  {'True':>5}  {'LIME Top-3':<50}  {'SHAP Top-3':<50}  Anchor Rule"
print(hdr); print("-"*len(hdr))
for idx in np.concatenate([bad_ids[:3], good_ids[:3]]):
    true  = "Good" if y_test.iloc[idx]==1 else "Bad"
    lime3 = " | ".join(get_lime_top(int(idx)))
    shap3 = " | ".join(get_shap_top(int(idx)))
    anc   = get_anchor_str(int(idx))
    print(f"{idx:>5}  {true:>5}  {lime3:<50}  {shap3:<50}  {anc}")


# ─────────────────────────────────────────────────────────────────────────────
# CELL 30 — FINAL SUMMARY
# ─────────────────────────────────────────────────────────────────────────────

mean_stab = stab_df["mean_rho"].mean()
final_df  = pd.DataFrame(all_metrics).set_index("model")\
             .sort_values("auc_roc", ascending=False)

print(f"""
╔════════════════════════════════════════════════════════════════════╗
║   PAPER IMPROVEMENT SUMMARY — arXiv:2103.00949 Extended           ║
╚════════════════════════════════════════════════════════════════════╝

ORIGINAL PAPER (Misheva et al. 2021)
  Dataset   : Lending Club (2.2M rows)
  Models    : LR, XGBoost, RF, SVM, Neural Network
  XAI       : LIME + SHAP (Tree, Kernel, Linear, Deep)
  Extras    : ALE plots for LR/SVM/NN
  Gaps      : No stability test, no faithfulness test,
              no counterfactuals, no rule-based explanations,
              no inter-method comparison metric

THIS WORK (Improvements)
  Dataset   : German Credit (standard benchmark, 1000 samples)
  Models    : All paper models + LightGBM + Optuna-tuned XGBoost
  XAI       : All paper methods + Anchors + DiCE (counterfactuals)

  QUANTITATIVE IMPROVEMENTS
  ─────────────────────────────────────────────────────────────────
  Tree vs Kernel SHAP consistency  : ρ = {rho_tk:.3f}
    (Paper §5.1.1 said 'consistent' but never measured it)

  SHAP vs Tree importance agreement : ρ = {rho_st:.3f}
    (Paper §5.3 compared visually; we quantify)

  SHAP vs LIME agreement            : ρ = {rho_sl:.3f}
    (NEW — paper never compared these two methods)

  LIME stability across seeds       : ρ = {mean_stab:.3f}
    (NEW — paper §7 noted LIME limitations but didn't measure)

  Faithfulness score                : {sum(r-s for s,r in zip(fa,ra)):.4f}
    (NEW — proves SHAP explanations reflect model behaviour)

  SMOTE effect on SHAP rankings     : ρ = {rho_sm:.3f}
    (NEW — does training balance change model's feature reliance?)

  Cost-sensitive score used         : Yes (German Credit cost matrix)
    (Paper reported only accuracy/F1/ROC)
""")

print("===== FINAL RESULTS TABLE =====")
print(final_df[["accuracy","balanced_acc","f1_bad",
                "auc_roc","auc_pr_bad","cost"]].to_string())
```
